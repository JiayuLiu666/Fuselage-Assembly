```python
import numpy as np

def get_fixed_features_from_z(z, threshold=1.0):
    z = np.asarray(z).reshape(-1)
    idx = np.where(np.abs(z) < threshold)[0]
    fixed_features = {int(i): 0 for i in idx}
    return fixed_features
```

```python
z = np.array([1e-2, 1e-3, 1e-4])
fixed_features = get_fixed_features_from_z(z, threshold=1.0)
print(fixed_features)   # {0: 0, 1: 0, 2: 0}
```
