```python
def _get_deviation(initPos, displacements, targetPos):
    '''
    Calculate the distance of the current node positions from their ideal positions
    ''' 
    finalPos = initPos + displacements
    deviations = finalPos - targetPos[:,0:2]
    return deviations.flatten()
```
